```python
import pandas as pd
import os
from datetime import datetime
import calendar

# =====================================================================
# GLOBAL STATE PROFILE STORAGE
# =====================================================================
# This master pool keeps track of cumulative entries across multiple inputs
global_ledger = []

def get_month_temporal_data():
    """Calculates live chronological metrics for the current calendar month."""
    today = datetime.now().date()
    current_day = today.day
    _, total_days_in_month = calendar.monthrange(today.year, today.month)
    days_remaining = (total_days_in_month - current_day) + 1
    return current_day, total_days_in_month, days_remaining


# =====================================================================
# MODULE 1: CONTINUOUS MANUAL DATA INGESTION ENGINE
# =====================================================================
def append_manual_expenses():
    print("\n--- 📂 MODULE 1: Manual Data Ingestion Activated ---")
    print("Enter transaction details. Type 'done' for the amount when finished.\n")

    added_count = 0
    while True:
        amount_input = input("1. Enter expense amount (e.g., 24.50) or type 'done': ")
        if amount_input.lower().strip() == 'done':
            print(f"\n✅ Data entry session complete. Appended {added_count} rows to master ledger.")
            break

        try:
            amount = float(amount_input)
            if amount <= 0:
                print("⚠️ Error: Expense must be greater than $0.00.\n")
                continue
        except ValueError:
            print("⚠️ Error: Invalid entry. Please enter a numerical price.\n")
            continue

        description = input("2. What was this expense for? (e.g., Coffee, Groceries): ").strip()

        while True:
            category_raw = input("3. Category? (Must type 'Essential' or 'Non-Essential'): ")
            category_clean = category_raw.lower().strip()
            if category_clean in ["essential", "non-essential", "non essential"]:
                category = "Non-Essential" if "non" in category_clean else "Essential"
                break
            else:
                print(f"⚠️ Validation Error: '{category_raw}' is invalid. Type 'Essential' or 'Non-Essential'.\n")

        merchant = input("4. Merchant/Store Name? (e.g., Woolworths, Mr Price): ").strip()

        while True:
            date_input = input("5. Enter expense date (YYYY-MM-DD) or press Enter for today: ").strip()
            if date_input == "":
                tx_date = datetime.now().date()
                break
            try:
                tx_date = datetime.strptime(date_input, "%Y-%m-%d").date()
                break
            except ValueError:
                print("⚠️ Formatting Error: Date must match the YYYY-MM-DD standard format (e.g., 2026-05-15).\n")

        # Build standard record structure
        transaction_record = {
            "Date": tx_date,
            "Amount": amount,
            "Description": description,
            "Category": category,
            "Merchant": merchant
        }

        # Directly push into our global state tracker!
        global_ledger.append(transaction_record)
        added_count += 1
        print(f"Logged: ${amount:.2f} spent at {merchant} on {tx_date}\n")


# ==========================================
# MODULE 4: CONTINUOUS CSV DATA INGESTION ENGINE
# ==========================================
def append_csv_expenses(file_path):
    print("\n--- 📂 MODULE 4: CSV Automated Ingestion Activated ---")
    if not os.path.exists(file_path):
        print(f"⚠️ System Error: The file '{file_path}' was not found.")
        print("   Please check the filename or upload it to your Colab side panel.")
        return

    try:
        df = pd.read_csv(file_path)
        df.columns = [col.strip() for col in df.columns]

        df['Category'] = df['Category'].astype(str).str.lower().str.strip()
        df['Merchant'] = df['Merchant'].astype(str).str.strip()
        df['Description'] = df['Description'].astype(str).str.strip()
        df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')
        df = df.dropna(subset=['Amount'])
        df = df[df['Amount'] > 0]

        if 'Date' not in df.columns:
            df['Date'] = datetime.now().date()
        else:
            df['Date'] = pd.to_datetime(df['Date'], errors='coerce').dt.date
            df = df.dropna(subset=['Date'])

        added_count = 0
        # Parse spreadsheet rows into our standardized global ledger
        for _, row in df.iterrows():
            category = "Non-Essential" if row['Category'] != 'essential' else "Essential"

            transaction_record = {
                "Date": row['Date'],
                "Amount": float(row['Amount']),
                "Description": row['Description'],
                "Category": category,
                "Merchant": row['Merchant']
            }
            global_ledger.append(transaction_record)
            added_count += 1

        print(f"📈 Spreadsheet successfully stacked! Appended {added_count} rows to master ledger.")

    except Exception as e:
        print(f"⚠️ Structural Ingestion Error: Failed to parse CSV file. Details: {e}")


# =====================================================================
# MODULE 5: CURRENT STATE LEDGER DISPLAY VISUALIZER
# =====================================================================
def print_current_ledger_summary():
    """Prints a rolling update of items currently locked inside the ledger memory pool."""
    if not global_ledger:
        print("\n=======================================================")
        print("📭 ROLLING MASTER LEDGER STATUS: Currently Empty")
        print("=======================================================")
        return

    print("\n=======================================================")
    print(f"          💰 ROLLING MASTER LEDGER POOL ({len(global_ledger)} Items Loaded)")
    print("=======================================================")
    for index, tx in enumerate(global_ledger, 1):
        print(f"{index}. {tx['Date']} | ${tx['Amount']:.2f} | {tx['Merchant']} | {tx['Description']} [{tx['Category']}]")
    print("=======================================================")


# =====================================================================
# MODULE 2: REFACTORED BUSINESS ANALYTICS SYSTEM
# =====================================================================
def process_final_financial_health(budget_amount, available_money, username):
    print("\n--- 📊 MODULE 2: Compiling Advanced Financial Metrics ---")

    if not global_ledger:
        print("⚠️ Computation Error: Ledger is empty. Cannot compile a report without data.")
        return

    # Extract metrics out of our shared state pool
    total_expenses = 0.0
    essential_amount = 0.0
    non_essential_amount = 0.0
    essential_count = 0
    non_essential_count = 0
    dates_list = []

    for tx in global_ledger:
        amt = tx["Amount"]
        cat = tx["Category"]
        dates_list.append(tx["Date"])

        total_expenses += amt
        if cat == "Essential":
            essential_amount += amt
            essential_count += 1
        else:
            non_essential_amount += amt
            non_essential_count += 1

    net_balance = available_money - total_expenses
    burn_rate = (total_expenses / budget_amount * 100)
    drawdown = (total_expenses / available_money * 100)

    essential_share_of_spending = (essential_amount / total_expenses * 100) if total_expenses > 0 else 0.0
    non_essential_share_of_spending = (non_essential_amount / total_expenses * 100) if total_expenses > 0 else 0.0

    target_savings_reserve = available_money * 0.20
    savings_rule_achieved = net_balance >= target_savings_reserve

    # Compute Time Metrics
    days_spanned = (max(dates_list) - min(dates_list)).days + 1
    curr_day, total_days, days_remaining = get_month_temporal_data()
    daily_burn_velocity = total_expenses / days_spanned
    safe_daily_allowance = (net_balance / days_remaining) if (days_remaining > 0 and net_balance > 0) else 0.0

    # Print Final Statement Report Document
    print("\n=======================================================================")
    print(f"        ✨ FINAL FINANCIAL HEALTH STATEMENT FOR: {username.upper()} ✨        ")
    print("=======================================================================")
    print(f" 🏁 Opening Balance (Available Real Cash) : ${available_money:.2f}")
    print(f" 🎯 Targeted Operational Budget Cap     : ${budget_amount:.2f}")
    print("-----------------------------------------------------------------------")
    print(f" 💸 CUMULATIVE EXPENDITURE VOLUME ({len(global_ledger)} Total Rows Combined):")
    print(f"    Total Capital Consumed              : ${total_expenses:.2f} ({drawdown:.1f}% of total cash used)")
    print(f"    ├── 🟢 Essential Funds Spent        : ${essential_amount:.2f} ({essential_count} transactions)")
    print(f"    └── 🔴 Non-Essential Funds Spent    : ${non_essential_amount:.2f} ({non_essential_count} transactions)")
    print("-----------------------------------------------------------------------")
    print(" 📊 INTERNAL COST COMPOSITION ANALYSIS   :")
    print(f"    ├── 🟢 Essential Allocation         : {essential_share_of_spending:.1f}% of spent dollars")
    print(f"    └── 🔴 Non-Essential Allocation     : {non_essential_share_of_spending:.1f}% of spent dollars")
    print("-----------------------------------------------------------------------")
    print(" ⏳ CALENDAR RUNWAY & VELOCITY ANALYTICS :")
    print(f"    Active Data Capture Window          : {days_spanned} days spanned across entries.")
    print(f"    Days Remaining in Current Month     : {days_remaining} day(s) left.")
    print(f"    🚀 Historical Spending Velocity     : ${daily_burn_velocity:.2f} consumed per calendar day.")
    print(f"    🛡️ Safe Remainder Daily Allowance   : ${safe_daily_allowance:.2f} maximum per remaining day.")
    print("-----------------------------------------------------------------------")
    print(" 🎯 BUDGET TARGET COMPLIANCE ALERTS :")
    if total_expenses == budget_amount:
        print(f"    📢 Status: \"Congrats, you spent exactly your budget!\"")
    elif total_expenses < budget_amount:
        surplus_savings = budget_amount - total_expenses
        print(f"    📢 Status: 🎉 \"Congratulations! You spent under your budget.\"")
        print(f"               You have an extra ${surplus_savings:.2f} to save or spend wisely!")
    else:
        overspending = total_expenses - budget_amount
        print(f"    📢 Status: ⚠️ \"Alert: You spent above your budget!\"")
        print(f"               You have gone over your target cap by ${overspending:.2f}.")
    print("-----------------------------------------------------------------------")
    print(" 🎯 80/20 BUDGET ALLOCATION STRATEGY BENCHMARK :")
    if savings_rule_achieved:
        print(f"    🏆 Benchmark Status: 🎉 SUCCESS! You successfully saved 20% or more, {username}!")
    else:
        shortfall = target_savings_reserve - net_balance
        print(f"    📉 Benchmark Status: ⚠️ SHORTFALL. You missed your 20% savings goal by ${shortfall:.2f}.")
    print("-----------------------------------------------------------------------")
    print(f" 🏁 CLOSING METRICS: Remaining Cash Left: ${net_balance:.2f} | Burn Rate: {burn_rate:.1f}%")
    print("=======================================================================")


# =====================================================================
# MODULE 3: PERSISTENT MAIN INTERFACE CONTROLLER
# =====================================================================
print("=======================================================")
print("  ✨ WELCOME TO YOUR PERSONAL BUDGET BUDDY ASSISTANT ✨  ")
print("=======================================================")

user_name = input("Hi there! What is your name? ").strip()
print(f"\nWelcome, {user_name}! Let's set up your core balances first.")
print("-------------------------------------------------------")

while True:
    cash_input = input(f"SETUP: Enter your starting Available Money (e.g., 2000): ")
    try:
        user_liquid_cash = float(cash_input)
        if user_liquid_cash <= 0:
            print("⚠️ Account Check: Balance must be greater than $0.00.\n")
            continue
        break
    except ValueError:
        print("⚠️ Formatting Error: Enter clean numerical values only.\n")

while True:
    budget_input = input(f"SETUP: Enter your Target Monthly Budget Cap (e.g., 500): ")
    try:
        user_budget_goal = float(budget_input)
        if user_budget_goal <= 0:
            print("⚠️ Budget Check: Target cap must be greater than $0.00.\n")
            continue
        break
    except ValueError:
        print("⚠️ Formatting Error: Enter clean numerical values only.\n")

# THE REVOLVING ACTION LOOP: Keeps running indefinitely until option 4 is hit!
while True:
    print_current_ledger_summary()
    print("\n=======================================================")
    print("               ⚙️ BUDGET BUDDY CONTROL CENTER         ")
    print("=======================================================")
    print(" [1] : 📝 Add expenses manually step-by-step")
    print(" [2] : 📎 Upload and merge an automated CSV file")
    print(" [3] : 🗑️  Clear ledger (Wipe out all inputs to restart)")
    print(" [4] : 📊 FINALIZE and compile my Financial Health Report")
    print("=======================================================")

    choice = input("Select an action code (1-4): ").strip()

    if choice == "1":
        append_manual_expenses()
    elif choice == "2":
        csv_filename = input("Enter file name (or press enter for 'expenses.csv'): ").strip()
        if csv_filename == "":
            csv_filename = "expenses.csv"
        append_csv_expenses(csv_filename)
    elif choice == "3":
        confirm = input("⚠️ Are you sure you want to completely clear all loaded entries? (yes/no): ").lower().strip()
        if confirm == 'yes':
            global_ledger.clear()
            print("\n🗑️ Master ledger tracking pool successfully wiped clean!")
    elif choice == "4":
        if not global_ledger:
            print("\n⚠️ Processing Denied: You cannot compile a statement with 0 expenses loaded.")
            print("   Please log some manual items or inject a CSV file first!\n")
            continue

        # Compile all entries sitting in state
        process_final_financial_health(user_budget_goal, user_liquid_cash, user_name)
        print(f"\nHope this report brought total clarity to your data today, {user_name}! 👋\n")
        break # Exit the app smoothly
    else:
        print("⚠️ Selection Error: Choose a clear option number from 1 to 4.\n")
```

```python
import gradio as gr
import pandas as pd
from datetime import datetime
import calendar
# Import the exact function verified in your class debugging handout
from hands_on_ai.chat import get_response

# =====================================================================
# SYSTEM BACKEND ANALYTICS ENGINE
# =====================================================================
def get_month_temporal_data():
    today = datetime.now().date()
    current_day = today.day
    _, total_days_in_month = calendar.monthrange(today.year, today.month)
    return current_day, total_days_in_month, (total_days_in_month - current_day) + 1

def add_manual_to_state(ledger, amount, desc, category, merchant, date_str):
    if not amount or amount <= 0: return ledger, update_visual_dataframe(ledger), "⚠️ Budget Buddy can't log $0 expenses."
    if not desc: return ledger, update_visual_dataframe(ledger), "⚠️ Please let your buddy know what this was for!"
    try: tx_date = datetime.strptime(date_str, "%Y-%m-%d").date() if date_str else datetime.now().date()
    except ValueError: tx_date = datetime.now().date()
    record = {"Date": tx_date, "Amount": float(amount), "Description": desc, "Category": category, "Merchant": merchant if merchant else "General Store"}
    updated_ledger = ledger + [record]
    return updated_ledger, update_visual_dataframe(updated_ledger), "✅ Logged successfully! Hit Finalize below when ready."

def upload_csv_to_state(ledger, file_obj):
    if file_obj is None: return ledger, update_visual_dataframe(ledger), "⚠️ No file picked up."
    try:
        df = pd.read_csv(file_obj.name)
        df.columns = [col.strip() for col in df.columns]
        df['Category'] = df['Category'].astype(str).str.lower().str.strip()
        df['Merchant'] = df['Merchant'].astype(str).str.strip()
        df['Description'] = df['Description'].astype(str).str.strip()
        df['Amount'] = pd.to_numeric(df['Amount'], errors='coerce')
        df = df.dropna(subset=['Amount'])
        df = df[df['Amount'] > 0]
        if 'Date' not in df.columns: df['Date'] = datetime.now().date()
        else: df['Date'] = pd.to_datetime(df['Date'], errors='coerce').dt.date; df = df.dropna(subset=['Date'])
        new_records = []
        for _, row in df.iterrows():
            cat_display = "Essential" if row['Category'] == 'essential' else "Non-Essential"
            new_records.append({"Date": row['Date'], "Amount": float(row['Amount']), "Description": row['Description'], "Category": cat_display, "Merchant": row['Merchant']})
        updated_ledger = ledger + new_records
        return updated_ledger, update_visual_dataframe(updated_ledger), f"📈 Merged {len(new_records)} items seamlessly!"
    except Exception as e: return ledger, update_visual_dataframe(ledger), f"⚠️ Spreadsheet parse issue: {e}"

def update_visual_dataframe(ledger):
    if not ledger: return pd.DataFrame(columns=["Date", "Amount", "Merchant", "Description", "Category"])
    return pd.DataFrame(ledger)

def compile_metrics(ledger, cash_input, budget_input):
    """Helper function to cleanly calculate data structures for the AI context."""
    if not ledger or not cash_input or not budget_input:
        return None
    total_expenses = sum(tx["Amount"] for tx in ledger)
    ess_amt = sum(tx["Amount"] for tx in ledger if tx["Category"] == "Essential")
    non_ess_amt = sum(tx["Amount"] for tx in ledger if tx["Category"] != "Essential")
    net_balance = cash_input - total_expenses
    return {
        "total_spent": total_expenses,
        "net_balance": net_balance,
        "essential_spent": ess_amt,
        "non_essential_spent": non_ess_amt,
        "budget_limit": budget_input
    }

def generate_cozy_statement(ledger, cash_input, budget_input, username):
    metrics = compile_metrics(ledger, cash_input, budget_input)
    if not metrics: return "### ⚠️ Hey Friend! Please complete steps 1 & 2 before Finalizing, okay?"

    burn_rate = (metrics["total_spent"] / budget_input * 100)
    target_savings_reserve = cash_input * 0.20
    savings_rule_achieved = metrics["net_balance"] >= target_savings_reserve
    _, _, days_remaining = get_month_temporal_data()
    safe_daily_allowance = (metrics["net_balance"] / days_remaining) if (days_remaining > 0 and metrics["net_balance"] > 0) else 0.0

    greeting = f"### 🦉 Okay {username if username else 'Friend'}, Here is Your Budget Buddy Review!\n\n"
    summary_table = f"| Starting Cash | Spending Goal | Total Spent | Net Left |\n| --- | --- | --- | --- |\n| **${cash_input:,.2f}** | **${budget_input:,.2f}** | **${metrics['total_spent']:,.2f}** | **${metrics['net_balance']:,.2f}** |\n\n"
    alerts = f"### ⏳ Calendar Runway & Velocity\n* To preserve your remaining **${metrics['net_balance']:,.2f}** through the month ({days_remaining} days left), your safe limit is **${safe_daily_allowance:,.2f} per day**.\n\n"
    compliance = "### 🎯 Target Goal Compliance\n" + (f"✨ Congrats! Exactly under your ${budget_input:,.2f} limit!" if metrics["total_spent"] <= budget_input else f"⚠️ Slipped over budget by **${(metrics['total_spent'] - budget_input):,.2f}**.") + "\n\n"
    allocation = f"### 📊 Cost Shares\n* Essentials: **${metrics['essential_spent']:,.2f}** | Non-Essentials: **${metrics['non_essential_spent']:,.2f}**\n\n"
    savings_check = "### 🎯 80/20 Strategy Review\n" + (f"🏆 **Success!** Retained over 20% savings (**${target_savings_reserve:,.2f}**)." if savings_rule_achieved else f"📉 **Shortfall.** Missed your 20% savings target of **${target_savings_reserve:,.2f}** by **${(target_savings_reserve - metrics['net_balance']):,.2f}**.") + "\n\n"
    return greeting + summary_table + alerts + compliance + allocation + savings_check

# =====================================================================
# HANDS-ON-AI INTERACTIVE CHAT ADVISOR PIPELINE
# =====================================================================
def budget_buddy_chat_response(user_message, chat_history, ledger, cash_input, budget_input, username):
    """
    Feeds financial data into the hands-on-ai prompt context, handles campus
    server string return flags gracefully, and keeps your chat running.
    """
    if not user_message:
        return "", chat_history

    metrics = compile_metrics(ledger, cash_input, budget_input)
    name = username if username else "Friend"

    if metrics:
        data_context = f"""
        User Name: {name}
        Current Financial Summary:
        - Starting Available Cash: ${cash_input:.2f}
        - User's Target Spend Cap: ${budget_input:.2f}
        - Total Amount Spent So Far: ${metrics['total_spent']:.2f}
        - Remaining Liquid Cash: ${metrics['net_balance']:.2f}
        - Breakdown: Spent ${metrics['essential_spent']:.2f} on Essentials and ${metrics['non_essential_spent']:.2f} on Non-Essentials.
        """
    else:
        data_context = f"User Name: {name}\nNo financial numbers have been recorded in the app ledger yet."

    system_prompt = f"""You are 'Budget Buddy', a helpful, warm, and comforting AI financial advisor.
    Use the following real-time ledger metadata to answer the user's questions accurately and warmly.
    Keep advice highly personalized, encouraging, practical, and clear.

    Context:
    {data_context}
    """

    try:
        # Request a response from the university server using assignment specs
        ai_reply = get_response(
            prompt=user_message,
            system=system_prompt,
            personality="friendly"
        )

        # INTERCEPTION GATEWAY: Catch if the server returned an error text string
        if "Connection error" in str(ai_reply) or "Error:" in str(ai_reply):
            raise ConnectionError("Campus server container capacity reached.")

    except Exception as e:
        # Cozy educational fallback logic so the student application never breaks down
        ai_reply = (
            f"☕ Hey {name}! I can see our campus server is a bit sleepy right now due to traffic. "
            f"Don't worry, your data is safe! Based on my offline calculations, you have "
            f"**${metrics['net_balance']:,.2f}** remaining in liquid funds. Let's try sending that message "
            f"again in a brief moment!" if metrics else
            f"☕ Hey {name}! The campus server is currently heavily loaded. Try hitting submit again in a few seconds!"
        )

    # Update and return the visual chat history elements safely
    chat_history.append((user_message, ai_reply))
    return "", chat_history

# =====================================================================
# REAL-TIME VISUAL INJECTION ENGINE
# =====================================================================
def apply_realtime_styles(color_preset, font_style):
    if font_style == "Casual Buddy (Serif)": font_family = "Georgia, serif"
    elif font_style == "Modern Minimalist (Mono)": font_family = "Courier New, monospace"
    else: font_family = "Inter, Arial, sans-serif"

    if color_preset == "rose (Spring Blossom Pink)":
        bg_color, card_color, btn_color, banner_text = "#fef1f3", "#fcc6d6", "#ef3167", "#4d0014"
    elif color_preset == "legolas (Organic Green)":
        bg_color, card_color, btn_color, banner_text = "#f2f7e0", "#c8da7f", "#587638", "#1a2408"
    elif color_preset == "glacier (Calming Blue)":
        bg_color, card_color, btn_color, banner_text = "#f0f4f9", "#ccdbff", "#153c67", "#031121"
    elif color_preset == "winnie the pooh (Yellow-Orange)":
        bg_color, card_color, btn_color, banner_text = "#fffdf0", "#ffd670", "#ff8c00", "#4d2b00"
    else: # cocoa-butter
        bg_color, card_color, btn_color, banner_text = "#f0e6da", "#dfab6e", "#8b4513", "#3d1e0a"

    style_html = f"""
    <style>
        .gradio-container, body, html {{ background-color: {bg_color} !important; font-family: {font_family} !important; }}
        .cozy-banner {{ background-color: {card_color} !important; border-radius: 12px; padding: 20px; text-align: center; }}
        .cozy-banner h1 {{ color: {banner_text} !important; margin: 0; }}
        .cozy-banner p {{ color: {banner_text} !important; margin: 5px 0 0 0; opacity: 0.8; }}
        button.primary {{ background-color: {btn_color} !important; color: white !important; border: none !important; font-weight: bold !important; border-radius: 8px !important; }}
        button.primary:hover {{ opacity: 0.9 !important; }}
    </style>
    """
    return style_html, f"✨ Mood adjusted seamlessly to '{color_preset}'!"

# =====================================================================
# GRADIO APPLICATION FRAMEWORK
# =====================================================================
with gr.Blocks(title="Budget Buddy Dashboard") as app:
    state_ledger = gr.State([])
    style_injection_box = gr.HTML("")

    # Visual Theme Customizer
    with gr.Row():
        gr.Markdown("## ☕ Customize Your Personal Budget Buddy Space")
    with gr.Row():
        color_preset = gr.Radio(
            choices=["cocoa-butter (Warm Brown)", "rose (Spring Blossom Pink)", "legolas (Organic Green)", "glacier (Calming Blue)", "winnie the pooh (Yellow-Orange)"],
            value="cocoa-butter (Warm Brown)", label="Pick a Cozy Color Vibe"
        )
        font_style = gr.Dropdown(
            choices=["Friendly & Rounded (Sans-Serif)", "Casual Buddy (Serif)", "Modern Minimalist (Mono)"],
            value="Friendly & Rounded (Sans-Serif)", label="Pick Your Typography Style"
        )
    apply_styles_btn = gr.Button("🎨 Apply Theme Styles", variant="secondary")
    gr.Markdown("---")

    # App Banner
    with gr.Row():
        gr.HTML('<div class="cozy-banner"><h1>☕ Your Personal Budget Buddy</h1><p>Let\'s look over your monthly metrics together, step-by-step!</p></div>')

    # Steps 1 & 2 Panels
    with gr.Row():
        with gr.Column():
            gr.Markdown("### 🌟 Step 1: Initialization")
            name_input = gr.Textbox(placeholder="What name can I call you?", label="Your First Name")
            cash_input = gr.Number(value=2000, label="💰 Available Real Cash ($)")
            budget_input = gr.Number(value=500, label="🎯 Target Spend Cap ($)")
        with gr.Column():
            gr.Markdown("### 📊 Step 2: Input Channels")
            with gr.Tabs():
                with gr.TabItem("📝 Manual Entry"):
                    tx_amt = gr.Number(label="Expense Cost ($)")
                    tx_desc = gr.Textbox(label="Description")
                    tx_cat = gr.Dropdown(choices=["Essential", "Non-Essential"], value="Non-Essential", label="Category")
                    tx_merch = gr.Textbox(label="Merchant Name")
                    tx_date = gr.Textbox(label="Date (YYYY-MM-DD)", value=datetime.now().strftime("%Y-%m-%d"))
                    manual_btn = gr.Button("➕ Log Record", variant="primary")
                with gr.TabItem("📎 CSV Bank Upload"):
                    file_upload = gr.File(file_types=[".csv"], label="Upload bank statements")
                    csv_btn = gr.Button("🚀 Scan and Merge Spreadsheet Rows", variant="primary")

    # Data Monitoring
    with gr.Row():
        with gr.Column():
            status_ticker = gr.Textbox(value="📭 Ledger empty...", label="Buddy Status Updates", interactive=False)
            ledger_view = gr.DataFrame(datatype=["str", "number", "str", "str", "str"], value=pd.DataFrame(columns=["Date", "Amount", "Merchant", "Description", "Category"]), interactive=False)
            clear_btn = gr.Button("🗑️ Clear Everything", variant="stop")

    # Steps 3 & 4 (Analytics + Live AI Chatbot)
    with gr.Row():
        with gr.Column():
            gr.Markdown("### 🦉 Step 3: Run Strategic Review")
            finalize_btn = gr.Button("📊 Compile My Financial Health Report!", variant="primary")
            report_box = gr.Markdown("✨ *Your statement will appear here.*")

        with gr.Column():
            gr.Markdown("### 💬 Step 4: Chat with Your Budget Buddy Advisor")
            chatbot_interface = gr.Chatbot(label="Budget Buddy Consultation")
            user_msg_input = gr.Textbox(placeholder="Ask me anything: 'Am I spending too much on lattes?' or 'How can I meet my savings goal?'", label="Type your message here...")
            chat_send_btn = gr.Button("💬 Ask Advisor", variant="primary")

    # Wiring Layout Events
    apply_styles_btn.click(fn=apply_realtime_styles, inputs=[color_preset, font_style], outputs=[style_injection_box, status_ticker])
    manual_btn.click(fn=add_manual_to_state, inputs=[state_ledger, tx_amt, tx_desc, tx_cat, tx_merch, tx_date], outputs=[state_ledger, ledger_view, status_ticker]).then(fn=lambda: (None, "", "Non-Essential", ""), outputs=[tx_amt, tx_desc, tx_cat, tx_merch])
    csv_btn.click(fn=upload_csv_to_state, inputs=[state_ledger, file_upload], outputs=[state_ledger, ledger_view, status_ticker])
    clear_btn.click(fn=lambda: ([], update_visual_dataframe([]), "🗑️ Wiped clean!"), outputs=[state_ledger, ledger_view, status_ticker])
    finalize_btn.click(fn=generate_cozy_statement, inputs=[state_ledger, cash_input, budget_input, name_input], outputs=[report_box])

    # Wire the new Chatbot interactive mechanism
    chat_send_btn.click(
        fn=budget_buddy_chat_response,
        inputs=[user_msg_input, chatbot_interface, state_ledger, cash_input, budget_input, name_input],
        outputs=[user_msg_input, chatbot_interface]
    )

app.launch(debug=True)
```
